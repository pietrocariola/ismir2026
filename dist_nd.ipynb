```python
import pandas as pd
import numpy as np
import transformations as tf
import os
from pathlib import Path
```

```python
input_path = '/home/gpu3/visfma/embeds'
file_name = 'x_fmasmallx_ele_014869_wav2clip_identity_none.npy'
transformation = 'pitchshift'
track = file_name.split('_')[2]+'_'+file_name.split('_')[3]
dataset = file_name.split('_')[1]
model = file_name.split('_')[4]

params = ['id']
for param in tf.tf_dict_params[transformation]:
    params.append(str(param)[:5])
df = pd.DataFrame(
    columns=params,
    index=params
)

for i, param1 in enumerate(params):
    if param1 == 'id':
        f1 = f"x_{dataset}_{track}_{model}_identity_none.npy"
        f1 = os.path.join(input_path, f1)
    else:
        f1 = f"x_{dataset}_{track}_{model}_{transformation}_{str(param1)[:5].replace('.', 'p')}.npy"
        f1 = os.path.join(input_path, f1)
    a1 = np.load(f1, 'r')[0,:]
    a1 = a1 / np.linalg.norm(a1)
    for j in range(i, len(params)):
        param2 = params[j]
        if param2 == 'id':
            f2 = f"x_{dataset}_{track}_{model}_identity_none.npy"
            f2 = os.path.join(input_path, f2)
        else:
            f2 = f"x_{dataset}_{track}_{model}_{transformation}_{str(param2)[:5].replace('.', 'p')}.npy"
            f2 = os.path.join(input_path, f2)
        a2 = np.load(f2, 'r')[0,:]
        a2 = a2 / np.linalg.norm(a2)
        d = np.sqrt(np.sum((a2-a1)**2))
        df.iloc[i,j] = d
        df.iloc[j,i] = d

df.to_csv("dist_nd.csv", index=True)
df.head(len(params))
```
